### Create a string, sent = ‘Coronavirus Caused Lockdowns Around The World.”

Convert each word in the list to lower case and store it in variable words_lower.

```python
words_lower = [word.lower() for word in words] 
print(words_lower)
```

### Create colors = [‘red’,’blue’,’orange’], and fruits = [‘orange’,’grapes’,’apples’] 

```python
colors_set = set()
fruits_set = set()
fruits_set.update({'orange', 'grapes', 'apples'})
```

```python
colors_set.update({'red', 'blue', 'orange'})
print('Color set is: ',colors_set)
print('Fruits set is: ',fruits_set)
```

Print color and fruits

Create colors_set, and fruits_set. (using set() ) and print them

```python
result = colors_set | fruits_set
print("Union of both set is: ",result)
```

```python

result = colors_set.intersection(fruits_set)
print("Intersection of both set is: ",result)
```

Find the intersection of both the sets 

```python

result = fruits_set.difference(colors_set)
print("Elements which are Fruits but not colors: ",result)
```
